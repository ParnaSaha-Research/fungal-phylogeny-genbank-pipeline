```python
%%writefile scripts/phylogenetic_analysis.py
import os
from Bio import AlignIO
from Bio.Phylo.TreeConstruction import DistanceCalculator, DistanceTreeConstructor
from Bio import Phylo
import pandas as pd
import subprocess

def run_phylogeny_pipeline():
    input_fasta = "data/fungal_its_contigs.fasta"
    output_dir = "results/phylogeny"
    os.makedirs(output_dir, exist_ok=True)

    aligned_fasta = os.path.join(output_dir, "its_aligned.fasta")
    tree_nwk = os.path.join(output_dir, "its_tree.nwk")
    distance_csv = os.path.join(output_dir, "pairwise_genetic_distances.csv")

    # 1. Run MAFFT for Multiple Sequence Alignment
    print("Step 1: Running Multiple Sequence Alignment with MAFFT...")
    mafft_cmd = f"mafft --auto {input_fasta} > {aligned_fasta}"
    subprocess.run(mafft_cmd, shell=True, check=True)
    print(f"Alignment saved to {aligned_fasta}")

    # 2. Calculate Pairwise Genetic Distances using Biopython
    print("Step 2: Calculating pairwise genetic distances...")
    alignment = AlignIO.read(aligned_fasta, "fasta")
    calculator = DistanceCalculator('identity')
    distance_matrix = calculator.get_distance(alignment)

    names = [seq.id for seq in alignment]
    dist_df = pd.DataFrame(distance_matrix.matrix, index=names, columns=names)
    dist_df.to_csv(distance_csv)
    print(f"Pairwise distance matrix saved to {distance_csv}")

    # 3. Run FastTree to build the Phylogenetic Tree
    print("Step 3: Constructing phylogenetic tree with FastTree...")
    fasttree_cmd = f"FastTree -nt -gtr {aligned_fasta} > {tree_nwk}"
    subprocess.run(fasttree_cmd, shell=True, check=True)
    print(f"Phylogenetic tree saved to {tree_nwk}")

    print("Phylogenetic pipeline completed successfully!")

if __name__ == "__main__":
    run_phylogeny_pipeline()
```

```python
from Bio import Phylo
import matplotlib.pyplot as plt

# Load the tree
tree_path = "results/phylogeny/its_tree.nwk"
tree = Phylo.read(tree_path, "newick")

# Set up a clean, high-resolution figure
fig, ax = plt.subplots(figsize=(12, 16), dpi=300)

# Draw the tree with custom aesthetics
Phylo.draw(
    tree,
    axes=ax,
    do_show=False,
    branch_labels=None,  # Keep branches clean; FastTree branch supports can be added if needed
    label_colors=dict()
)

# Customize plot styling
plt.title("Phylogenetic Analysis of Fungal Isolates (ITS Region)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Evolutionary Distance", fontsize=11, fontweight='semibold')

# Clean up graph borders for a professional look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#cccccc')
ax.spines['bottom'].set_color('#cccccc')
ax.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()

# Save publication-quality image
output_img = "results/phylogeny/phylogenetic_tree_publication.png"
plt.savefig(output_img, dpi=300, bbox_inches='tight')
plt.show()

print(f"Publication-quality tree saved successfully to {output_img}!")
```

```python
%%writefile scripts/its2_structure_prediction.py
import os
import subprocess
import shutil
from Bio import SeqIO

def predict_its2_structures():
    input_fasta = "data/fungal_its_contigs.fasta"
    output_dir = "results/its2_structures"
    os.makedirs(output_dir, exist_ok=True)

    # Find RNAfold path automatically
    rnafold_path = shutil.which("RNAfold")
    if not rnafold_path:
        # Check common micromamba/conda paths
        possible_paths = ["/root/micromamba/bin/RNAfold", "/opt/conda/bin/RNAfold", "/usr/local/bin/RNAfold"]
        for p in possible_paths:
            if os.path.exists(p):
                rnafold_path = p
                break

    if not rnafold_path:
        raise FileNotFoundError("RNAfold executable not found! Please check the installation.")

    print(f"Using RNAfold at: {rnafold_path}")
    print("Running ITS2 secondary structure prediction with RNAfold...")

    output_file = os.path.join(output_dir, "its2_fold_results.txt")

    cmd = f"{rnafold_path} < {input_fasta} > {output_file}"
    subprocess.run(cmd, shell=True, check=True)

    print(f"Structure predictions saved successfully to {output_file}!")

if __name__ == "__main__":
    predict_its2_structures()
```

```python
import shutil
import os

for folder in ['data', 'results', 'scripts']:
    src = f'/content/{folder}'
    dst = f'/content/fungal-phylogeny-genbank-pipeline/{folder}'
    if os.path.exists(src) and os.path.isdir(src):
        if not os.path.exists(dst):
            shutil.move(src, dst)
        else:
            for item in os.listdir(src):
                s = os.path.join(src, item)
                d = os.path.join(dst, item)
                if os.path.isdir(s):
                    shutil.copytree(s, d, dirs_exist_ok=True)
                else:
                    shutil.copy2(s, d)

print("All folders successfully consolidated inside fungal-phylogeny-genbank-pipeline!")
```
